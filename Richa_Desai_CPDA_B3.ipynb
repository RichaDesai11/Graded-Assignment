### 1) For the suburb of Altona, it is postulated that a typical property sells for $800000.

Use the data at hand to test this assumption.
Is the typical property price really $800,000 or has it increased?
Use a significance level of 5%.

```python
# One-Sample T-Test
Altona = df[df['Suburb'] == 'Altona']['Price']

# H0: Altona Property Average Price is $800,000.
#H1: Altona Property Average Price is more than $800,000. 

alpha = 0.05

t_stat, p_value = ttest_1samp(Altona, 800000)

print("Mean:", Altona.mean())
print("T-stat:", t_stat)
print ("P-value:", p_value)

if p_value < alpha:
    print("Reject Null Hypothesis") 
else: 
    print("Fail to Reject Null Hypothesis")
```

### 2) For the year 2016, is there any difference in the prices of properties sold in the summer months vs winter months?

Consider months from October till March as winter months and rest as summer months.
Use a significance level of 5%. 

```python
# Convert Date column to datetime
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)

# Extracting year and month
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month

# Filtering data for year 2016
df_2016 = df[df['Year'] == 2016]

# Defining seasons
# Winter: October (10) to March (3)
# Summer: April (4) to September (9)

winter_months = [10, 11, 12, 1, 2, 3]
summer_months = [4, 5, 6, 7, 8, 9]

# Creating separate datasets
winter_prices = df_2016[df_2016['Month'].isin(winter_months)]['Price']
summer_prices = df_2016[df_2016['Month'].isin(summer_months)]['Price']

# Descriptive statistics
print("Winter Months Statistics")
print("Sample Size:", len(winter_prices))
print("Mean Price:", winter_prices.mean())
print("Standard Deviation:", winter_prices.std())

print("\nSummer Months Statistics")
print("Sample Size:", len(summer_prices))
print("Mean Price:", summer_prices.mean())
print("Standard Deviation:", summer_prices.std())

# H0: Average Property Price in Winter is same as Average Property Price in Summer
# H1: Average Property Price in Winter is not same as Average Property Price in Summer

# Performing Independent two-sample t-test
t_stat, p_value = ss.ttest_ind(winter_prices, summer_prices, equal_var=False)

# Print results
print("\nT-Test Results")
print("T-statistic:", t_stat)
print("P-value:", p_value)

# Significance level
alpha = 0.05

# Decision
if p_value < alpha:
    print("\nReject the null hypothesis")
    print("There is a significant difference in property prices between winter and summer months in 2016.")
else:
    print("\nFail to reject the null hypothesis")
    print("There is no significant difference in property prices between winter and summer months in 2016.")
```

### 6) One-Sample Hypothesis Test (Industry Pricing) A real estate firm claims that the average property price in Richmond is $1,000,000. Using the dataset, test whether the actual average price is significantly different from this claim at a 5% significance level.

```python
# Filter data for Richmond suburb
richmond = df[df['Suburb'] == 'Richmond']

# Extract property prices and remove missing values
prices = richmond['Price'].dropna()

# Claimed population mean
mu = 1000000

# Descriptive statistics
sample_size = len(prices)
sample_mean = prices.mean()
sample_std = prices.std()

# H0: Average Property Price in Richmond is $1,000,000
# H1: Average Property Price in Rachmond is not equal to $1,000,000

# Perform one-sample t-test
t_stat, p_value = ttest_1samp(prices, mu)

# Display statistics
print("Richmond Property Price Analysis")
print("--------------------------------")
print("Sample Size:", sample_size)
print("Sample Mean:", sample_mean)
print("Sample Standard Deviation:", sample_std)

# Display hypothesis test results
print("\nHypothesis Test Results")
print("-----------------------")
print("T-statistic:", t_stat, 3)
print("P-value:", p_value, 5)

# Significance level
alpha = 0.05

# Decision
print("\nDecision")
print("---------")

if p_value < alpha:
    print("Reject the null hypothesis")
    print("The average property price in Richmond is significantly different from $1,000,000.")
else:
    print("Fail to reject the null hypothesis")
    print("There is not enough evidence to conclude that the average property price differs from $1,000,000.")
```

### 7) Independent Two-Sample T-Test (Feature Impact) 
### Do properties with car parking sell at a higher average price than properties without car parking, across the entire dataset? Use a 5% significance level.

```python
# Create two groups
# Group 1: Properties WITH car parking
with_parking = df[df['Car'] > 0]['Price']

# Group 2: Properties WITHOUT car parking
without_parking = df[df['Car'] == 0]['Price']

# Descriptive statistics
print("Properties WITH Car Parking")
print("----------------------------")
print("Sample Size:", len(with_parking))
print("Average Price:", with_parking.mean())
print("Standard Deviation:", with_parking.std(), 2)

print("\nProperties WITHOUT Car Parking")
print("-------------------------------")
print("Sample Size:", len(without_parking))
print("Average Price:", without_parking.mean())
print("Standard Deviation:", without_parking.std())

# H0: Properties with parking have Same Prices
# H1: Properties with parking have Higher prices

# Perform independent two-sample t-test
t_stat, p_value = ss.ttest_ind(with_parking, without_parking, equal_var=False)

# Display test results
print("\nT-Test Results")
print("----------------")
print("T-statistic:", t_stat)
print("Two-tailed p-value:", p_value)

# Significance level
alpha = 0.05

# Decision
print("\nDecision")
print("----------")

if p_value < alpha:
    print("Reject the null hypothesis")
    print("There is insufficient evidence that parking increases property prices.")
else:
    print("Fail to reject the null hypothesis")
    print("Properties with car parking sell at significantly higher prices.")
```

### 8) Two-Way ANOVA (Location & Property Type) 
Investigate whether property prices are influenced by:
• Suburb
• Type of property
• Interaction between suburb and property type

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Convert categorical variables into category datatype
df['Suburb'] = df['Suburb'].astype('category')
df['Type'] = df['Type'].astype('category')

# Build Two-Way ANOVA model
# Includes:
# 1. Main effect of Suburb
# 2. Main effect of Type
# 3. Interaction effect between Suburb and Type

model = ols('Price ~ C(Suburb) + C(Type) + C(Suburb):C(Type)',data=df).fit()

# Generate ANOVA table
anova_table = sm.stats.anova_lm(model, typ=2)

# Display ANOVA results
print("\nTwo-Way ANOVA Results")
print("----------------------")
print(anova_table)

# Significance level
alpha = 0.05

print("\nInterpretation")
print("---------------")

# Check significance of Suburb
if anova_table.loc['C(Suburb)', 'PR(>F)'] < alpha:
    print("Suburb significantly affects property prices.")
else:
    print("Suburb does NOT significantly affect property prices.")

# Check significance of Property Type
if anova_table.loc['C(Type)', 'PR(>F)'] < alpha:
    print("Property type significantly affects property prices.")
else:
    print("Property type does NOT significantly affect property prices.")

# Check significance of Interaction
interaction_term = 'C(Suburb):C(Type)'

if anova_table.loc[interaction_term, 'PR(>F)'] < alpha:
    print("There is a significant interaction between suburb and property type.")
else:
    print("There is NO significant interaction between suburb and property type.")
```

### 10)Industry-Style Hypothesis Validation (Policy Decision) 
A housing policy group believes that properties with more than 2 bathrooms command a premium price. 
Design and execute a statistical test to validate this claim: 
Identify the correct test 
• State hypotheses
• Report p-value 
• Give a clear recommendation to policymakers 

```python
# Create two groups

# Group 1: Properties with MORE than 2 bathrooms
more_than_2 = df[df['Bathroom'] > 2]['Price']

# Group 2: Properties with 2 or fewer bathrooms
two_or_less = df[df['Bathroom'] <= 2]['Price']

# Display descriptive statistics
print("Properties with MORE than 2 Bathrooms")
print("--------------------------------------")
print("Sample Size:", len(more_than_2))
print("Average Price:", more_than_2.mean())
print("Standard Deviation:", more_than_2.std())

print("\nProperties with 2 or Fewer Bathrooms")
print("--------------------------------------")
print("Sample Size:", len(two_or_less))
print("Average Price:", two_or_less.mean())
print("Standard Deviation:", two_or_less.std())

# Perform Independent Two-Sample T-Test
# Welch's t-test

# H0: Average Price is not affected by the number of bathrooms
# H1: Average Price is more when number of bathroom increases

t_stat, p_value = ss.ttest_ind(more_than_2,two_or_less, equal_var=False)

# Since the claim is:
# Properties with >2 bathrooms have HIGHER prices

# Display test results
print("\nHypothesis Test Results")
print("------------------------")
print("T-statistic:",t_stat)
print("Two-tailed p-value:", p_value)

# Significance level
alpha = 0.05

# Decision
print("\nDecision")
print("----------")

if p_value < alpha and t_stat > 0:
    print("Reject the null hypothesis.")
    print("Properties with more than 2 bathrooms command significantly higher prices")
else:
    print("Fail to reject the null hypothesis.")
    print("There is insufficient evidence that properties with more than 2 bathrooms have higher prices.")
```
